```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")  #n significa modelo nano

# Entrenamiento (fine-tuning)
results = model.train(
    data="C:/Users/sam20/OneDrive/Documentos/IA/CuartoIA/Proyecto_Integrador_2/Proyecto/Construction PPE.v3i.yolov8/data.yaml",  
    epochs=50,             # Número de épocas (ajusta según dataset)
    imgsz=640,             # Tamaño de entrada
    batch=16,              # Batch size (reduce si tienes poca memoria)
    lr0=0.001,             # Learning rate inicial
    optimizer="SGD",       # O 'Adam' para convergencia más suave
    device=0,              # Usa 0 si tienes GPU CUDA; si no, pon 'cpu'
    name="ppe_yolov8_finetuned",  # Carpeta donde se guardan resultados
    pretrained=True,       # Usa pesos preentrenados de COCO
    workers=4              # Número de procesos de carga de datos
)

# Validación (opcional, pero útil)
metrics = model.val()
print(metrics)

# Exportar el modelo a otro formato
model.export(format="onnx")  # o 'torchscript', 'engine', etc.

```
